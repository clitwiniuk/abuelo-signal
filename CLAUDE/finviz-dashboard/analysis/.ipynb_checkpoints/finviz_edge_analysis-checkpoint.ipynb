# Finviz Dashboard — Edge Analysis

Análisis de patrones con edge a partir de los snapshots de Finviz capturados cada 15 min.

**DB activa:** `~/Library/Application Support/finviz-dashboard/finviz_snapshots.db`  
**Período:** 2026-03-24 → presente  
**Categorías disponibles:** Top Gainers, New High, Unusual Volume, Insider Buying, Upgrades, Earnings Before/After, etc.

```python
import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from pathlib import Path
from collections import Counter

DB_PATH = Path.home() / 'Library/Application Support/finviz-dashboard/finviz_snapshots.db'
print(f'DB: {DB_PATH}')
print(f'Existe: {DB_PATH.exists()}')

conn = sqlite3.connect(DB_PATH)
df = pd.read_sql('SELECT * FROM snapshots', conn)
conn.close()

# Parsear tipos
df['timestamp'] = pd.to_datetime(df['timestamp'], utc=True).dt.tz_convert('US/Eastern')
df['date'] = df['timestamp'].dt.date
df['hour'] = df['timestamp'].dt.hour
df['minute'] = df['timestamp'].dt.minute
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df['change_pct'] = pd.to_numeric(df['change_pct'].str.replace('%','', regex=False), errors='coerce')
df['volume'] = pd.to_numeric(df['volume'], errors='coerce')

print(f'\nTotal filas: {len(df):,}')
print(f'Período: {df["date"].min()} → {df["date"].max()}')
print(f'\nCategorías:')
print(df.groupby('category').size().sort_values(ascending=False).to_string())
```

## 1. Frecuencia de aparición por ticker

¿Qué tickers aparecen más veces y en qué categorías? La persistencia en listas de Finviz puede indicar momentum real.

```python
# Top tickers por apariciones totales
ticker_counts = df.groupby('ticker').size().sort_values(ascending=False)
print('Top 30 tickers por apariciones totales:')
print(ticker_counts.head(30).to_string())

# Por categoría
cat_ticker = df.groupby(['category','ticker']).size().reset_index(name='count')
top_per_cat = cat_ticker.sort_values('count', ascending=False).groupby('category').head(5)
print('\nTop 5 por categoría:')
print(top_per_cat.to_string(index=False))
```

## 2. Co-aparición en múltiples listas (señal de confluencia)

Un ticker que aparece simultáneamente en `Top Gainers` + `Unusual Volume` + `Insider Buying` es una señal más fuerte.
Buscamos confluencias por timestamp (snapshot).

```python
# Agrupar por timestamp → ticker → set de categorías
# Usar ventanas de 15 min (un snapshot)
df['snapshot'] = df['timestamp'].dt.floor('15min')

confluences = (
    df.groupby(['snapshot', 'ticker'])['category']
    .apply(lambda x: frozenset(x))
    .reset_index()
)
confluences.columns = ['snapshot', 'ticker', 'categories']
confluences['n_lists'] = confluences['categories'].apply(len)

multi = confluences[confluences['n_lists'] >= 2].sort_values(['n_lists','snapshot'], ascending=[False,False])
print(f'Eventos con confluencia >= 2 listas: {len(multi)}')
print('\nTop 20:')
print(multi.head(20)[['snapshot','ticker','n_lists','categories']].to_string(index=False))
```

```python
# ¿Qué combinaciones de categorías son más frecuentes?
combo_counts = Counter(confluences[confluences['n_lists'] >= 2]['categories'])
print('Combinaciones más frecuentes (>= 2 listas):')
for combo, cnt in combo_counts.most_common(20):
    print(f'  {cnt:4d}x  {" + ".join(sorted(combo))}')
```

## 3. Persistencia en listas (días consecutivos)

¿Cuántos días seguidos aparece un ticker en una categoría? Mayor persistencia = momentum más sostenido.

```python
# Días únicos que aparece cada ticker en cada categoría
daily_presence = df.groupby(['ticker','category','date']).size().reset_index(name='snapshots_that_day')

# Número de días únicos
days_present = daily_presence.groupby(['ticker','category'])['date'].nunique().reset_index(name='days')
days_present = days_present.sort_values('days', ascending=False)

print('Ticker-categoría con más días distintos de aparición:')
print(days_present.head(30).to_string(index=False))
```

## 4. Insider Buying — análisis específico

Insider Buying es una de las categorías con más edge demostrado en estudios. Analizamos qué tickers han tenido insider buying y si después aparecen en otras listas.

```python
insider = df[df['category'] == 'Insider Buying'].copy()
print(f'Snapshots con Insider Buying: {len(insider)}')
print(f'Tickers únicos: {insider["ticker"].nunique()}')
print()

insider_tickers = set(insider['ticker'].unique())

# Ver si esos tickers aparecen después en Top Gainers
gainers = df[df['category'] == 'Top Gainers']
insider_then_gainers = gainers[gainers['ticker'].isin(insider_tickers)]

print(f'Tickers con Insider Buying que TAMBIÉN aparecieron en Top Gainers: {insider_then_gainers["ticker"].nunique()}')
print(insider_then_gainers.groupby('ticker').agg(
    veces_gainer=('ticker','count'),
    max_change=('change_pct','max'),
    avg_change=('change_pct','mean')
).sort_values('max_change', ascending=False).head(20).to_string())
```

## 5. Unusual Volume — detección de pre-breakout

Unusual Volume + New High o Top Gainers puede ser señal de breakout válido.

```python
# Tickers en Unusual Volume
uv = df[df['category'] == 'Unusual Volume']['ticker'].unique()
nh = df[df['category'] == 'New High']['ticker'].unique()
tg = df[df['category'] == 'Top Gainers']['ticker'].unique()

# Confluencias
uv_and_nh = set(uv) & set(nh)
uv_and_tg = set(uv) & set(tg)
uv_and_both = set(uv) & set(nh) & set(tg)

print(f'Tickers en Unusual Volume:          {len(uv)}')
print(f'Tickers en Unusual Volume + New High: {len(uv_and_nh)} → {sorted(uv_and_nh)}')
print(f'Tickers en Unusual Volume + Top Gainers: {len(uv_and_tg)} → {sorted(uv_and_tg)}')
print(f'Tickers en UV + New High + Top Gainers: {len(uv_and_both)} → {sorted(uv_and_both)}')
```

## 6. Análisis por hora del día — ¿cuándo aparecen en las listas?

¿Las capturas de apertura (9:30-10:00) son más predictivas que las del mediodía?

```python
fig, axes = plt.subplots(2, 1, figsize=(14, 8))

# Snapshots por hora
hourly = df.groupby(['hour','category']).size().unstack(fill_value=0)
cats_plot = ['Top Gainers', 'Unusual Volume', 'New High', 'Insider Buying', 'Upgrades']
cols = [c for c in cats_plot if c in hourly.columns]
hourly[cols].plot(kind='bar', ax=axes[0], width=0.8)
axes[0].set_title('Snapshots por hora del día (ET)')
axes[0].set_xlabel('Hora ET')
axes[0].set_ylabel('Número de registros')
axes[0].legend(loc='upper right', fontsize=8)

# Change % medio en Top Gainers por hora
tg_df = df[df['category'] == 'Top Gainers'].dropna(subset=['change_pct'])
tg_hour = tg_df.groupby('hour')['change_pct'].agg(['mean','median','count'])
tg_hour[['mean','median']].plot(kind='bar', ax=axes[1])
axes[1].set_title('Top Gainers — Change % medio y mediana por hora')
axes[1].set_xlabel('Hora ET')
axes[1].set_ylabel('Change %')

plt.tight_layout()
plt.show()
```

## 7. Tickers que aparecen el mismo día en múltiples categorías (watchlist diaria)

Esta sección genera una tabla por fecha con los tickers más relevantes del día.

```python
# Por día: tickers y cuántas categorías distintas tuvieron
daily_multi = (
    df.groupby(['date','ticker'])['category']
    .apply(lambda x: list(set(x)))
    .reset_index()
)
daily_multi.columns = ['date','ticker','categories']
daily_multi['n_cats'] = daily_multi['categories'].apply(len)
daily_multi = daily_multi.sort_values(['date','n_cats'], ascending=[False, False])

# Mostrar los últimos 5 días
last_dates = sorted(daily_multi['date'].unique())[-5:]
for d in reversed(last_dates):
    subset = daily_multi[daily_multi['date'] == d]
    print(f'\n=== {d} ===')
    print(subset[['ticker','n_cats','categories']].head(15).to_string(index=False))
```

## 8. Resumen de edge potencial

Combinaciones con más señal basadas en los datos disponibles.

```python
# Tabla resumen: tickers con confluencia alta (>= 3 categorías en algún día)
strong_signals = daily_multi[daily_multi['n_cats'] >= 3].copy()
strong_signals['cats_str'] = strong_signals['categories'].apply(lambda x: ' | '.join(sorted(x)))

# Enriquecer con precio y cambio del día
price_info = df.groupby(['date','ticker']).agg(
    avg_price=('price','mean'),
    max_change=('change_pct','max')
).reset_index()

strong_signals = strong_signals.merge(price_info, on=['date','ticker'], how='left')
strong_signals = strong_signals.sort_values(['date','n_cats'], ascending=[False,False])

print(f'Eventos con >= 3 categorías en el mismo día: {len(strong_signals)}')
print()
print(strong_signals[['date','ticker','n_cats','avg_price','max_change','cats_str']]
      .to_string(index=False))
```
