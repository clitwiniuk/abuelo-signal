# Intraday First Appearance — Edge Analysis

**Hipótesis:** Cuando un ticker aparece POR PRIMERA VEZ en Top Gainers de Finviz con un cambio del **20-40%**, 
tiene una probabilidad muy alta de continuar subiendo hasta el cierre del mismo día.

**Origen:** Descubierto el 2026-04-09 analizando 11 días de snapshots Finviz (Mar 24 – Abr 9).

**Lógica de mercado:**
- Un ticker con 20-40% al aparecer todavía está en movimiento. El momentum está activo.
- Un ticker con >60% ya está en fase de distribución. El dinero inteligente está saliendo.

**DB activa:** `~/Library/Application Support/finviz-dashboard/finviz_snapshots.db`  
**Resolución Finviz:** snapshot cada ~15 minutos durante RTH  
**Exit proxy:** último snapshot del día (no el cierre real — ver sección IBKR más abajo)

```python
import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from pathlib import Path
from scipy import stats

DB_PATH = Path.home() / 'Library/Application Support/finviz-dashboard/finviz_snapshots.db'
print(f'DB: {DB_PATH}')
print(f'Existe: {DB_PATH.exists()}')

conn = sqlite3.connect(DB_PATH)
df = pd.read_sql("SELECT * FROM snapshots WHERE category='Top Gainers'", conn)
conn.close()

# Parse timestamp y fecha
df['timestamp'] = pd.to_datetime(df['timestamp'], utc=True).dt.tz_convert('US/Eastern')
df['date'] = df['timestamp'].dt.date
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df['change_pct'] = pd.to_numeric(df['change_pct'].str.replace('%','', regex=False), errors='coerce')

# Parse volumen con sufijos M/K (ej: "133.71M", "42.02M", "152.99K")
def parse_volume(v):
    if pd.isna(v) or v == '' or v == '0': return np.nan
    v = str(v).strip()
    try:
        if v.endswith('B'): return float(v[:-1]) * 1e9
        elif v.endswith('M'): return float(v[:-1]) * 1e6
        elif v.endswith('K'): return float(v[:-1]) * 1e3
        else: return float(v)
    except:
        return np.nan

df['volume'] = df['volume'].apply(parse_volume)

print(f'\nTotal snapshots Top Gainers: {len(df):,}')
print(f'Período: {df["date"].min()} → {df["date"].max()}')
print(f'Días de mercado: {df["date"].nunique()}')
print(f'Tickers únicos: {df["ticker"].nunique()}')
print(f'Registros con volumen: {df["volume"].notna().sum()} / {len(df)}')
```

## 1. Construcción del dataset de trades

Para cada ticker-día:
- **Entry**: primer snapshot del día con change_pct > 0
- **Exit**: último snapshot del mismo día
- **Filtros de calidad**: al menos 3 snapshots en el día, al menos 60 minutos entre entry y exit

```python
# Filtrar solo datos con cambio real
df_valid = df[df['change_pct'] > 0].copy()

# Por ticker-día: primer y último snapshot
first = (
    df_valid.sort_values('timestamp')
    .groupby(['ticker','date'])
    .first()
    .reset_index()
    .rename(columns={'timestamp':'entry_ts','price':'entry_price','change_pct':'entry_chg','volume':'entry_vol'})
)

last = (
    df_valid.sort_values('timestamp')
    .groupby(['ticker','date'])
    .last()
    .reset_index()
    .rename(columns={'timestamp':'exit_ts','price':'exit_price','change_pct':'exit_chg'})
)

# Contar snapshots por ticker-día
n_snaps = df_valid.groupby(['ticker','date']).size().reset_index(name='n_snaps')

# Unir
trades = first[['ticker','date','entry_ts','entry_price','entry_chg','entry_vol']].merge(
    last[['ticker','date','exit_ts','exit_price','exit_chg']], on=['ticker','date']
).merge(n_snaps, on=['ticker','date'])

# Calcular duración y retorno
trades['duration_min'] = (trades['exit_ts'] - trades['entry_ts']).dt.total_seconds() / 60
trades['trade_ret'] = (trades['exit_price'] - trades['entry_price']) / trades['entry_price'] * 100
trades['entry_hour'] = trades['entry_ts'].dt.hour + trades['entry_ts'].dt.minute / 60

# Filtros de calidad
trades_filtered = trades[
    (trades['n_snaps'] >= 3) &
    (trades['duration_min'] >= 60)
].copy()

print(f'Ticker-días antes de filtro: {len(trades)}')
print(f'Ticker-días después de filtro (n>=3, dur>=60min): {len(trades_filtered)}')
print(f'\nResumen general:')
print(f'  N: {len(trades_filtered)}')
print(f'  Win Rate: {(trades_filtered["trade_ret"] > 0).mean():.1%}')
print(f'  Mean return: {trades_filtered["trade_ret"].mean():.2f}%')
print(f'  Median return: {trades_filtered["trade_ret"].median():.2f}%')
print(f'  Std dev: {trades_filtered["trade_ret"].std():.2f}%')
print(f'  Best: {trades_filtered["trade_ret"].max():.1f}% ({trades_filtered.loc[trades_filtered["trade_ret"].idxmax(), "ticker"]})')
print(f'  Worst: {trades_filtered["trade_ret"].min():.1f}% ({trades_filtered.loc[trades_filtered["trade_ret"].idxmin(), "ticker"]})')
```

## 2. Segmentación por change% en la primera aparición

Esta es la variable más importante. La hipótesis es que 20-40% es el sweet spot.

```python
# Buckets de change%
bins = [0, 20, 40, 60, 100, 999]
labels = ['<20%', '20-40%', '40-60%', '60-100%', '>100%']
trades_filtered['chg_bucket'] = pd.cut(trades_filtered['entry_chg'], bins=bins, labels=labels)

bucket_stats = trades_filtered.groupby('chg_bucket', observed=True).agg(
    N=('trade_ret','count'),
    WR=('trade_ret', lambda x: (x > 0).mean()),
    mean_ret=('trade_ret','mean'),
    median_ret=('trade_ret','median'),
    best=('trade_ret','max'),
    worst=('trade_ret','min')
).round(2)

print('Resultados por bucket de change% en primera aparición:')
print(bucket_stats.to_string())

# p-value para el bucket 20-40%
bucket_a = trades_filtered[trades_filtered['chg_bucket'] == '20-40%']['trade_ret']
t_stat, p_val = stats.ttest_1samp(bucket_a, 0)
from statsmodels.stats.proportion import proportion_confint
n_wins = (bucket_a > 0).sum()
ci_low, ci_high = proportion_confint(n_wins, len(bucket_a), method='wilson')
print(f'\nBucket 20-40% — test estadístico:')
print(f'  N={len(bucket_a)}, WR={n_wins/len(bucket_a):.1%}')
print(f'  IC 95% WR: [{ci_low:.1%}, {ci_high:.1%}]')
print(f'  p-value (t-test vs 0): {p_val:.4f}')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Win Rate por bucket
wr_vals = bucket_stats['WR'] * 100
colors = ['#2ecc71' if v >= 60 else '#e74c3c' for v in wr_vals]
axes[0].bar(wr_vals.index, wr_vals.values, color=colors, edgecolor='black', linewidth=0.5)
axes[0].axhline(50, color='black', linestyle='--', linewidth=1, label='50% (azar)')
axes[0].set_title('Win Rate por Change% al aparecer', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Win Rate (%)')
axes[0].set_ylim(0, 110)
for i, v in enumerate(wr_vals.values):
    axes[0].text(i, v + 1, f'{v:.0f}%', ha='center', fontweight='bold')
axes[0].legend()

# Median return por bucket
med_vals = bucket_stats['median_ret']
colors2 = ['#2ecc71' if v > 0 else '#e74c3c' for v in med_vals]
axes[1].bar(med_vals.index, med_vals.values, color=colors2, edgecolor='black', linewidth=0.5)
axes[1].axhline(0, color='black', linewidth=1)
axes[1].set_title('Retorno Mediano por Change% al aparecer', fontsize=13, fontweight='bold')
axes[1].set_ylabel('Retorno mediano (%)')
for i, v in enumerate(med_vals.values):
    axes[1].text(i, v + 0.3, f'{v:.1f}%', ha='center', fontweight='bold')

plt.suptitle('Finviz Top Gainers — Señal de primera aparición (intraday)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 3. Segmentación por volumen al aparecer

Hipótesis secundaria: volumen bajo al aparecer = más recorrido por delante.

```python
# Terciles de volumen — robusto a NaN
vol_clean = trades_filtered['entry_vol'].dropna()
q33 = vol_clean.quantile(0.33)
q66 = vol_clean.quantile(0.66)

def vol_bucket(v):
    if pd.isna(v) or v == 0: return 'Sin datos'
    elif v <= q33: return f'Bajo (<{q33/1e6:.1f}M)'
    elif v <= q66: return f'Medio ({q33/1e6:.1f}-{q66/1e6:.1f}M)'
    else: return f'Alto (>{q66/1e6:.1f}M)'

trades_filtered['vol_tercile'] = trades_filtered['entry_vol'].apply(vol_bucket)

vol_stats = trades_filtered[trades_filtered['vol_tercile'] != 'Sin datos'].groupby('vol_tercile').agg(
    N=('trade_ret','count'),
    WR=('trade_ret', lambda x: (x > 0).mean()),
    mean_ret=('trade_ret','mean'),
    median_ret=('trade_ret','median'),
    vol_median=('entry_vol', lambda x: x.median() / 1e6)
).round(2)

print(f'Terciles: Q33={q33/1e6:.1f}M, Q66={q66/1e6:.1f}M')
print(f'Sin datos de volumen: {(trades_filtered["vol_tercile"] == "Sin datos").sum()}')
print()
print('Resultados por volumen en primera aparición (terciles):')
print(vol_stats.to_string())
```

## 4. Segmentación por hora de primera aparición

```python
trades_filtered['time_bucket'] = pd.cut(
    trades_filtered['entry_hour'],
    bins=[0, 10.5, 12.0, 24],
    labels=['Early (<10:30)', 'Mid (10:30-12:00)', 'Late (>12:00)']
)

time_stats = trades_filtered.groupby('time_bucket', observed=True).agg(
    N=('trade_ret','count'),
    WR=('trade_ret', lambda x: (x > 0).mean()),
    mean_ret=('trade_ret','mean'),
    median_ret=('trade_ret','median')
).round(2)

print('Resultados por hora de primera aparición:')
print(time_stats.to_string())
```

## 5. Combinaciones de filtros

Buscamos la combinación que maximiza WR con N >= 10.

```python
combos = []
chg_buckets = [b for b in trades_filtered['chg_bucket'].cat.categories]
vol_buckets = [b for b in trades_filtered['vol_tercile'].unique() if b != 'Sin datos']
time_buckets = [b for b in trades_filtered['time_bucket'].cat.categories]

for chg_b in chg_buckets:
    for vol_b in [None] + vol_buckets:
        for time_b in [None] + list(time_buckets):
            mask = trades_filtered['chg_bucket'] == chg_b
            if vol_b: mask &= trades_filtered['vol_tercile'] == vol_b
            if time_b: mask &= trades_filtered['time_bucket'] == time_b
            subset = trades_filtered[mask]
            if len(subset) >= 10:
                combos.append({
                    'chg': chg_b, 'vol': vol_b or 'all', 'time': time_b or 'all',
                    'N': len(subset),
                    'WR': (subset['trade_ret'] > 0).mean(),
                    'mean_ret': subset['trade_ret'].mean(),
                    'median_ret': subset['trade_ret'].median()
                })

combos_df = pd.DataFrame(combos).sort_values('WR', ascending=False)
print('Top combinaciones (N >= 10), ordenadas por Win Rate:')
print(combos_df.head(15).round(3).to_string(index=False))
```

## 6. Trades individuales — mejores y peores

```python
cols = ['ticker','date','entry_ts','entry_price','exit_price','trade_ret','entry_chg','entry_vol','duration_min']

print('TOP 10 MEJORES TRADES:')
print(trades_filtered.nlargest(10, 'trade_ret')[cols].round(2).to_string(index=False))

print('\nTOP 10 PEORES TRADES:')
print(trades_filtered.nsmallest(10, 'trade_ret')[cols].round(2).to_string(index=False))
```

## 7. La señal principal — detalle del bucket 20-40%

Análisis completo del subconjunto con edge demostrado.

```python
signal = trades_filtered[trades_filtered['chg_bucket'] == '20-40%'].copy()

print(f'=== SEÑAL: Primera aparición con 20-40% ===')
print(f'N: {len(signal)}')
print(f'Win Rate: {(signal["trade_ret"] > 0).mean():.1%}')
print(f'Mean return: {signal["trade_ret"].mean():.2f}%')
print(f'Median return: {signal["trade_ret"].median():.2f}%')
print(f'Std dev: {signal["trade_ret"].std():.2f}%')

print(f'\nDistribución de retornos:')
print(signal['trade_ret'].describe().round(2))

print(f'\nTodos los trades del bucket 20-40%:')
print(signal[cols].sort_values('trade_ret', ascending=False).round(2).to_string(index=False))

# Distribución visual
fig, ax = plt.subplots(figsize=(12, 4))
colors = ['#2ecc71' if r > 0 else '#e74c3c' for r in signal['trade_ret']]
signal_sorted = signal.sort_values('trade_ret', ascending=False)
ax.bar(range(len(signal_sorted)), signal_sorted['trade_ret'], color=colors, edgecolor='black', linewidth=0.3)
ax.axhline(0, color='black', linewidth=1)
ax.set_xticks(range(len(signal_sorted)))
ax.set_xticklabels(
    [f"{r['ticker']}\n{str(r['date'])[5:]}" for _, r in signal_sorted.iterrows()],
    rotation=45, ha='right', fontsize=7
)
ax.set_title('Retornos individuales — Señal 20-40% primera aparición', fontsize=13, fontweight='bold')
ax.set_ylabel('Retorno intraday (%)')
green_patch = mpatches.Patch(color='#2ecc71', label='Winner')
red_patch = mpatches.Patch(color='#e74c3c', label='Loser')
ax.legend(handles=[green_patch, red_patch])
plt.tight_layout()
plt.show()
```

## 8. Validación con datos IBKR reales

El exit proxy de Finviz es el último snapshot (~15:45-15:57 ET), no el cierre real.
Aquí descargamos los daily bars de IBKR para los tickers del bucket 20-40% y comparamos.

**Requisito:** TWS abierto con clientId disponible.

```python
# Descarga de datos IBKR para validación
# Ejecutar solo con TWS activo

import sys
sys.path.insert(0, str(Path.home() / 'proyectos/TRADING/1_PROYECTOS_ACTIVOS/CLAUDE/trading_system_v3'))

try:
    from ib_insync import IB, Stock
    import asyncio

    ib = IB()
    ib.connect('127.0.0.1', 7497, clientId=99, readonly=True)
    print(f'Conectado a TWS: {ib.isConnected()}')

    signal_tickers = signal[['ticker','date','entry_price','entry_chg','trade_ret']].copy()
    results = []

    for _, row in signal_tickers.iterrows():
        try:
            contract = Stock(row['ticker'], 'SMART', 'USD')
            bars = ib.reqHistoricalData(
                contract,
                endDateTime=f"{row['date']} 23:59:59 US/Eastern",
                durationStr='3 D',
                barSizeSetting='1 day',
                whatToShow='TRADES',
                useRTH=True
            )
            if bars:
                day_bar = next((b for b in bars if str(b.date) == str(row['date'])), None)
                if day_bar:
                    ibkr_ret = (day_bar.close - row['entry_price']) / row['entry_price'] * 100
                    results.append({
                        'ticker': row['ticker'],
                        'date': row['date'],
                        'entry_price': row['entry_price'],
                        'finviz_exit': row['trade_ret'],
                        'ibkr_close': day_bar.close,
                        'ibkr_ret': ibkr_ret,
                        'ibkr_high': day_bar.high,
                        'ibkr_low': day_bar.low,
                    })
        except Exception as e:
            print(f'  Error {row["ticker"]}: {e}')

    ib.disconnect()

    ibkr_df = pd.DataFrame(results)
    print(f'\nTickers con datos IBKR: {len(ibkr_df)}')
    print(f'\nComparación Finviz proxy vs cierre real IBKR:')
    print(f'  WR Finviz proxy: {(ibkr_df["finviz_exit"] > 0).mean():.1%}')
    print(f'  WR IBKR cierre: {(ibkr_df["ibkr_ret"] > 0).mean():.1%}')
    print(f'  Mean ret Finviz: {ibkr_df["finviz_exit"].mean():.2f}%')
    print(f'  Mean ret IBKR:   {ibkr_df["ibkr_ret"].mean():.2f}%')
    print()
    print(ibkr_df.round(2).to_string(index=False))

except Exception as e:
    print(f'TWS no disponible o error: {e}')
    print('Ejecutar con TWS activo para validar con datos reales.')
```

## 9. Resumen y estado de la hipótesis

Actualizar después de cada sesión de análisis.

```python
signal_a = trades_filtered[trades_filtered['chg_bucket'] == '20-40%']
n = len(signal_a)
wr = (signal_a['trade_ret'] > 0).mean()
ci_lo, ci_hi = proportion_confint((signal_a['trade_ret'] > 0).sum(), n, method='wilson')
t, p = stats.ttest_1samp(signal_a['trade_ret'], 0)

print('=' * 60)
print('ESTADO DE LA HIPÓTESIS')
print('=' * 60)
print(f'Hipótesis: Primera aparición en Top Gainers con 20-40%')
print(f'           tiene edge long intraday hasta el cierre.')
print()
print(f'Período analizado: {trades_filtered["date"].min()} → {trades_filtered["date"].max()}')
print(f'Días de mercado:   {trades_filtered["date"].nunique()}')
print()
print(f'N (bucket 20-40%): {n}')
print(f'Win Rate:          {wr:.1%}')
print(f'IC 95% WR:         [{ci_lo:.1%}, {ci_hi:.1%}]')
print(f'Median return:     {signal_a["trade_ret"].median():.2f}%')
print(f'Mean return:       {signal_a["trade_ret"].mean():.2f}%')
print(f'p-value (ret>0):   {p:.4f}')
print()
if n < 50:
    print('⚠ MUESTRA INSUFICIENTE — Necesitas N>=50 para conclusiones sólidas.')
    print(f'  Faltan aprox. {50-n} trades más (~{(50-n)//10} semanas de mercado).')
elif p < 0.05 and wr > 0.60:
    print('✓ SEÑAL CON EVIDENCIA — p<0.05 y WR>60%. Testear en live con size mínimo.')
else:
    print('? SEÑAL DÉBIL o NO SIGNIFICATIVA — continuar acumulando datos.')
print()
print('Limitaciones conocidas:')
print('  - Exit proxy = último snapshot Finviz, no cierre real (sección 8)')
print('  - Sin slippage modelado (spreads amplios en smallcaps)')
print('  - Sin filtro de precio mínimo (algunos tickers < $1)')
print('  - Survivor bias: tickers que desaparecen rápido de la lista excluidos')
print('=' * 60)
```
